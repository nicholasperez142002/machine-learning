# Section #3 - Plotting mulitple graphs on the same axes

## Simple

```python
fig, ax = plt.subplots()
ax.plot(x,x ** 3, label='x cubed')
ax.plot(x, y_damped_oscillation, label='damped oscillation')
ax.legend() # This allows the labels to show up
plt.show()
```

## Scatter plot, coloring by labels

```python
fig, ax = plt.subplots()

#First, plot the data where the individual does not have heart disease
df_subset = df[df['target'] == 0]
ax.scatter(df_subset['trestbps'].values, df_subset['chol'].values, label='healthy')
#Second, plot the data where the individual does have heart disease
df_subset = df[df['target'] == 1]
ax.scatter(df_subset['trestbps'].values, df_subset['chol'].values, label='heart disease')
ax.legend()
plt.show()
```

```python
df['target'].unique()
```

```python
label_mapping = {0: "healthy", 1: 'heart disease'}
```

```python
fig, ax = plt.subplots()
for target_value in df['target'].unique():
    df_subset = df[df['target'] == target_value]
    ax.scatter(df_subset['trestbps'].values, df_subset['chol'].values, label=label_mapping[target_value])

ax.legend()
plt.show()
```

## Scatterplot, calling out a single point

```python
fig, ax =  plt.subplots()

ax.scatter(df['trestbps'].values, df['chol'].values, c='lightgray')
ax.scatter(df['trestbps'].values[:10], df['chol'].values[:10], c='red')
plt.show()
```
